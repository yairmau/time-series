## A simulator you will reuse

`simulate_ar(phis, n, seed)` simulates an AR($p$) process, where `phis` is a list: `[phi1]` for AR(1), `[phi1, phi2]` for AR(2), and so on. White noise is `simulate_ar([], n, seed)`, and a random walk is the cumulative sum of white noise.

```python
def simulate_ar(phis, n, seed):
    """phis: list of AR coefficients, e.g. [0.7] for AR(1), [0.5, -0.8] for AR(2)"""
    rng = np.random.default_rng(seed)
    p = len(phis)
    eps = rng.normal(size=n)
    x = list(eps[:max(p, 1)])
    for t in range(len(x), n):
        x.append(eps[t] + sum(phi * x[t - 1 - i] for i, phi in enumerate(phis)))
    return pd.Series(x[:n])

seed = 123   # change it to the last digits of your ID number
```
