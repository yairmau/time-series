## Import data and basic processing

The station and year are the same as in the previous homeworks.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from statsmodels.tsa.seasonal import seasonal_decompose, STL, MSTL
```

```python
# write here the name of the .csv file you downloaded from the IMS
filename = "my_station.csv"

df = pd.read_csv(filename, na_values=['-'])

# the IMS column that holds the time starts with "Date & Time"
time_column = [c for c in df.columns if c.startswith('Date & Time')][0]
df['date'] = pd.to_datetime(df[time_column], dayfirst=True)
df = df.set_index('date')

# shorter names for the columns we will use
df = df.rename(columns={'Temperature (°C)': 'temperature',
                        'Rainfall (mm)': 'rain',
                        'Relative humidity (%)': 'humidity',
                        'Global radiation (W/m^2)': 'radiation'})
df.head()
```

Now we average the 10-minute data to hourly values, so that one day is exactly 24 data points. Decomposition methods need a value at every time step, so short gaps are filled by linear interpolation. Check the numbers below: if the month you choose for parts 1-3 has long gaps, choose another month.

```python
hourly_raw = df['temperature'].resample('h').mean()

missing = hourly_raw.isna()
gap_lengths = missing.groupby((~missing).cumsum()).sum()   # length of each run of missing hours
print("missing hours in the year:", missing.sum())
print("longest gap (hours):", gap_lengths.max())
print("missing hours per month:")
print(missing.groupby(missing.index.month).sum().to_string())

hourly = hourly_raw.interpolate()   # fill the gaps
```
